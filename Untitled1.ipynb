```python
def is_year_leap(year):
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    else:
        return False

year = int(input("Введите год: "))
print(is_year_leap(year))
```

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True

n = int(input("Введите число: "))
print(is_prime(n))
```

```python
a = float(input("Введите первую переменную: "))
b = float(input("Введите вторую переменную: "))

left = abs(-138 / 2) ** 1.3
right = abs((-69 / 28 ** 5.1) * 4)

if left > right:
    left, right = right, left

if a > b * 3.6 or left <= b <= right:
    a, b = b, a

print("a =", a)
print("b =", b)
```

```python
nums = []
for i in range(5):
    nums.append(int(input("Введите число: ")))

unique = True
for i in range(5):
    for j in range(i + 1, 5):
        if nums[i] == nums[j]:
            unique = False
even_count = 0
neg_count = 0
range_count = 0
for n in nums:
    if n % 2 == 0:
        even_count += 1
    if n < 0:
        neg_count += 1
    if -256 <= n <= 1024:
        range_count += 1

print("Все числа уникальны:", unique)
print("Количество чётных:", even_count)
print("Количество отрицательных:", neg_count)
print("Количество в промежутке [-256; 1024]:", range_count)
```
